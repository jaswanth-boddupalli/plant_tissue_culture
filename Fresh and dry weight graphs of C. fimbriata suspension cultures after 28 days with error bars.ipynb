```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data (Replace this with your actual data)
categories = ['Control', 'JA 50', 'JA 100', 'JA 200', 'JA 300']
mean_values = [3.28, 6.23, 8.15, 9.76, 10.36]
std_devs = [0.01, 0.06, 0.12, 0.18, 0.16]

# Define the colors (muted color for the mean points and error bars)
mean_color = "#1f77b4"

# Create a figure with specific size
fig, ax = plt.subplots(figsize=(8, 6))

# Error bars (Mean ± SD) with refined aesthetics
ax.errorbar(
    categories, mean_values, yerr=std_devs, fmt='o', 
    color=mean_color, ecolor=mean_color, elinewidth=1.5, capsize=6, capthick=1.5,
    markerfacecolor='white', markeredgewidth=1.5, markersize=8, zorder=3
)

# Customize plot for publication standards
ax.set_title('Mean ± SD for Different Groups', fontsize=14, weight='bold', pad=15)
ax.set_xlabel('Groups', fontsize=12, weight='bold', labelpad=10)
ax.set_ylabel('Values', fontsize=12, weight='bold', labelpad=10)
ax.set_xticks(range(len(categories)))
ax.set_xticklabels(categories, fontsize=11)
ax.set_yticks(np.arange(0, 45, 5))
ax.set_yticklabels(np.arange(0, 45, 5), fontsize=11)

# Remove spines to make the figure cleaner
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Add gridlines with soft appearance
ax.grid(True, which='both', linestyle='--', linewidth=0.5, alpha=0.5)

# Tight layout to prevent clipping
fig.tight_layout()

# Save as high-resolution PNG or TIFF
fig.savefig("mean_sd_refined_plot.png", dpi=300)
fig.savefig("mean_sd_refined_plot.tiff", dpi=300)

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data provided by the user
categories = ['Control', 'JA 50', 'JA 100', 'JA 200', 'JA 300']
mean_values_1 = [3.28, 6.23, 8.15, 9.82, 10.02]
std_devs_1 = [0.01, 0.06, 0.12, 0.12, 0.09]
mean_values_2 = [1.09, 2.34, 2.70, 3.26, 3.32]
std_devs_2 = [0.12, 0.27, 0.34, 0.43, 0.39]

# Define colors
color_1 = '#228B22'  # Green color
color_2 = '#1E90FF'  # Blue color

# Create a figure with specific size
fig, ax = plt.subplots(figsize=(8, 6))

# Convert categories to a range for plotting
x = np.arange(len(categories))

# Plot line with error bars for the first set of data
ax.errorbar(x, mean_values_1, yerr=std_devs_1, fmt='-o', 
            color=color_1, ecolor=color_1, elinewidth=1.5, capsize=4,
            markerfacecolor=color_1, markeredgewidth=1.5, markersize=6, label='Fresh Weight')

# Plot line with error bars for the second set of data
ax.errorbar(x, mean_values_2, yerr=std_devs_2, fmt='-s', 
            color=color_2, ecolor=color_2, elinewidth=1.5, capsize=4,
            markerfacecolor=color_2, markeredgewidth=1.5, markersize=6, label='Dry Weight')

# Customize plot for publication standards
ax.set_title('Mean ± SD for JA Treatments', fontsize=14, weight='bold', pad=15)
ax.set_xlabel('Jasmonic Acid Treatment (µM)', fontsize=12, weight='bold', labelpad=10)
ax.set_ylabel('Weight of callus biomass [g]', fontsize=12, weight='bold', labelpad=10)
ax.set_xticks(x)
ax.set_xticklabels(categories, fontsize=11)

# Set y-axis ticks to multiples of 2 (0, 2, 4, ..., 12)
ax.set_yticks(np.arange(0, 13, 2))

# Remove gridlines for a cleaner look
ax.grid(False)

# Remove unnecessary spines to make it look cleaner
for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)

# Add a legend
ax.legend(fontsize=11, loc='upper right', frameon=False)

# Tight layout to prevent clipping
fig.tight_layout()

# Show the plot
plt.show()
```
