```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
data = {
    'Medium': ['B5 Medium', 'WPM Medium', 'MS Medium'],
    '% multiple shoot induction': [53.3, 33.33, 86.66],
    '% callus induction': [60, 33.33, 93.33]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Melt the DataFrame for Seaborn
df_melted = df.melt(
    id_vars='Medium',
    var_name='Induction Type',
    value_name='Percentage'
)

# Set style and context
sns.set_context("paper", font_scale=1.2)
sns.set_style("white")  # Removes grid lines in the background

# Custom palette
custom_palette = sns.color_palette("colorblind")

# Create the figure, specifying DPI for better resolution
plt.figure(figsize=(6, 4), dpi=300)

# Plot the bar chart
chart = sns.barplot(
    data=df_melted, 
    x='Medium', 
    y='Percentage', 
    hue='Induction Type', 
    palette=custom_palette,
    edgecolor='black',
    saturation=0.8,
    dodge=True,  # Ensures bars of different categories are next to each other
    width=0.6    # Reduces bar width for less spacing between groups
)

# Annotate percentages on top of each bar
for bar, value in zip(chart.patches, df_melted['Percentage']):
    height = bar.get_height()  # Get the height of the bar
    chart.text(
        bar.get_x() + bar.get_width() / 2,  # X position (center of bar)
        height + 1,  # Y position (slightly above the bar)
        f'{value:.1f}%',  # Text to display (formatted percentage)
        ha='center',  # Align horizontally at the center of the bar
        va='bottom',  # Align vertically above the bar
        fontsize=10   # Font size of the annotation
    )

# Title and labels
plt.xlabel("Culture Medium", fontsize=12)
plt.ylabel("Percentage (%)", fontsize=12)

# Tick label formatting
plt.xticks(fontsize=10, rotation=0)
plt.yticks(fontsize=10)

# Legend
plt.legend(
    title="Induction Type",
    fontsize=10,
    title_fontsize=10,
    loc='best',
    frameon=True
)

# Remove top and right spines
sns.despine()

# Use tight_layout for better spacing
plt.tight_layout()

# Save or show
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
labels = [
    "IAA 0.1", "IAA 0.2", "IAA 0.3",
    "IBA 0.1", "IBA 0.2", "IBA 0.3",
    "NAA 0.1", "NAA 0.2", "NAA 0.3"
]

mean_roots = [4.23, 2.69, 1.18, 3.08, 1.56, 1.09, 12.06, 7.42, 3.91]
se_roots = [0.12, 0.17, 0.11, 0.08, 0.05, 0.09, 0.06, 0.24, 0.08]

mean_length = [1.83, 1.66, 1.71, 1.62, 1.86, 1.75, 3.00, 2.05, 1.33]
se_length = [0.03, 0.06, 0.03, 0.08, 0.02, 0.02, 0.02, 0.03, 0.07]

# X positions
x = np.arange(len(labels))
bar_width = 0.4

# Custom colors
color_roots = '#1f77b4'  # Blue for Number of Roots per Shoot
color_length = '#2ca02c'  # Green for Root Length

# Create figure and axis
fig, ax1 = plt.subplots(figsize=(12, 6))

# Plot Number of Roots per Shoot on the primary y-axis
bars_roots = ax1.bar(
    x - bar_width / 2, mean_roots, bar_width, yerr=se_roots, capsize=5,
    color=color_roots, alpha=0.9, label='Number of Roots per Shoot'
)
ax1.set_ylabel("Number of Roots per Shoot (Mean ± SE)", fontsize=12, color=color_roots)
ax1.tick_params(axis='y', labelcolor=color_roots)

# Create secondary y-axis for Root Length
ax2 = ax1.twinx()
bars_length = ax2.bar(
    x + bar_width / 2, mean_length, bar_width, yerr=se_length, capsize=5,
    color=color_length, alpha=0.9, label='Root Length (cm)'
)
ax2.set_ylabel("Root Length (cm, Mean ± SE)", fontsize=12, color=color_length)
ax2.tick_params(axis='y', labelcolor=color_length)

# X-axis labels
ax1.set_xticks(x)
ax1.set_xticklabels(labels, rotation=45, ha='right')
ax1.set_xlabel("Auxin Concentrations (mg/l)", fontsize=12)

# Title and legend

fig.legend(loc='upper left', bbox_to_anchor=(0.1, 1.0), fontsize=10)

# Remove grid lines
ax1.grid(False)
ax2.grid(False)

# Adjust layout for readability
plt.tight_layout()

# Show plot
plt.show()
```
